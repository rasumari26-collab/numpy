```python
Copy - 'Deep Copying'
This mehod makes a complete copy of an existing array and its elements.

When the newly created array is modified, it will not affect the existing array or vice versa.

    ]
```

```python
a=np.arange(3,12)
b=a.copy()
print('Original Array is:',a)
print('--'*30)
b[0]=16
print('Original Array is:',a)
print('--'*30)
print('Modified Array is:',b)
```

```python
view - 'Shallow copy'
Elements in the view modified will also modify the elements in the original array.

So, both the arrays will act as one and the same.
```

```python
a=np.arange(5,18)
b=a.view()
print('Original Array is:',a)
print('*'*50)
b[0]=16
print('Original Array is:',a)
print('*'*50)
print('Modified Array is:',b)
```

```python
a[1]=25
print('Original Array is:',a)
print('*'*50)
print('Modified Array is:',b)
```
